```python
def print_formatted(number):
    width = len(bin(number)[2:])

    for i in range(1, number + 1):
        decimal = str(i).rjust(width)
        octal = oct(i)[2:].rjust(width)
        hexa = hex(i)[2:].upper().rjust(width)
        binary = bin(i)[2:].rjust(width)

        print(decimal, octal, hexa, binary)

if __name__ == '__main__':
    n = int(input())
    print_formatted(n)
```

```python
def print_rangoli(size):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    lines = []
    width = 4 * size - 3

    for i in range(size):
        left = alphabet[size-1:size-i-1:-1]
        right = alphabet[size-i-1:size]

        row = "-".join(left + right)
        lines.append(row.center(width, "-"))

    print("\n".join(lines + lines[-2::-1]))



if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

```python
import math
import os
from datetime import datetime

def time_delta(t1, t2):
    a = datetime.strptime(t1, '%a %d %b %Y %H:%M:%S %z')
    b = datetime.strptime(t2, '%a %d %b %Y %H:%M:%S %z')

    diff = abs(a - b)
    seconds=int(diff.total_seconds())

    return str(seconds)


if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'],'w')

    t = int(input())

    for t_itr in range(t):
        t1 = input()
        t2 = input()

        delta = time_delta(t1,t2)
        fptr.write(delta+'\n')
```

```python
T=int(input())

def check(N):
    if len(N)==0:
        return False
    if N[0]=="+" or N[0]=="-":
        N=N[1:]
    if N.count(".")!=1:
        return False
    parti=N.split(".")
    prima=parti[0]
    dopo=parti[1]
    if len(dopo)==0:
        return False
    for e in prima:
        if not e.isdigit():
            return False
    for f in dopo:
        if not f.isdigit():
            return False
    return True

for i in range(T):
        N=input()
        print(check(N))
```

```python
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_starttag(self,tag,attrs):
        print("Start :",tag)
        for attr,value in attrs:
            print("->",attr,">",value)

    def handle_endtag(self,tag):
        print("End   :",tag)

    def handle_startendtag(self,tag,attrs):
        print("Empty :",tag)
        for attr,value in attrs:
            print("->",attr,">",value)

parser=MyHTMLParser()

for _ in range(int(input())):
    parser.feed(input())
```

```python
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_comment(self, data):
        if '\n' in data:
            print(">>> Multi-line Comment")
        else:
            print(">>> Single-line Comment")
        print(data)

    def handle_data(self, data):
        if data.strip():
            print(">>> Data")
            print(data)
html = ""
for i in range(int(input())):
    html += input().rstrip()
    html += '\n'

parser = MyHTMLParser()
parser.feed(html)
parser.close()
```

```python
import sys
import xml.etree.ElementTree as etree

def get_attr_number(node):
    count=len(node.attrib)
    for child in node:
        count+=get_attr_number(child)

    return count

if __name__ == '__main__':
    sys.stdin.readline()
    xml = sys.stdin.read()
    tree = etree.ElementTree(etree.fromstring(xml))
    root = tree.getroot()
    print(get_attr_number(root))
```

```python
import xml.etree.ElementTree as etree

maxdepth = 0
def depth(elem, level):
    global maxdepth
    level+=1
    maxdepth=max(maxdepth,level)
    for child in elem:
        depth(child,level)

if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

```python
def wrapper(f):
    def fun(l):
        numeri=[]
        for numero in l:
            numero=numero[-10:]
            numero="+91 "+numero[:5]+" "+numero[5:]
            numeri.append(numero)
        f(numeri)
    return fun

@wrapper
def sort_phone(l):
    print(*sorted(l), sep='\n')

if __name__ == '__main__':
    l = [input() for _ in range(int(input()))]
    sort_phone(l)
```

```python
import operator

def person_lister(f):
    def inner(people):
        people.sort(key=lambda x:int(x[2]))
        return [f(person) for person in people]
    return inner

@person_lister
def name_format(person):
    return ("Mr. " if person[3] == "M" else "Ms. ") + person[0] + " " + person[1]

if __name__ == '__main__':
    people = [input().split() for i in range(int(input()))]
    print(*name_format(people), sep='\n')
```
